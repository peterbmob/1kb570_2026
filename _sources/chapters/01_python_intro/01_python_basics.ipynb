## 1.1 Variables and Data Types

A **variable** is just a labelled box that holds a value so you can refer to it
by name instead of retyping the value everywhere. Python figures out the type
of box automatically from what you put in it — you never have to declare
"this is a number" the way some other languages require. The four most common
primitive types are:

| Type | Example | Use |
|------|---------|-----|
| `int` | `3` | Integer counts, indices |
| `float` | `3.14` | Measured values |
| `str` | `'ZnO'` | Labels, file names |
| `bool` | `True` | Flags, conditions |

Why bother distinguishing them at all? Because operations behave differently
depending on type — `3 / 2` gives a fractional result (`1.5`), string
concatenation (`'Li' + 'FePO4'`) glues text together rather than adding
numbers, and mixing an `int` count of atoms with a `float` measured mass is
exactly the distinction you want to keep straight when writing formulas
below.

````python
# ── Material properties as Python variables ──────────────────────────────────
material_name   = 'LiFePO4'          # str
density         = 3.6                # g/cm³  (float)
num_formula_units = 4                # int
is_cathode      = True               # bool

print(type(material_name), material_name)
print(type(density),       density)
print(type(num_formula_units), num_formula_units)
print(type(is_cathode),    is_cathode)
````

````python
# ── Basic arithmetic ─────────────────────────────────────────────────────────
molar_mass_Li  = 6.941    # g/mol
molar_mass_Fe  = 55.845
molar_mass_P   = 30.974
molar_mass_O   = 15.999

MM_LiFePO4 = molar_mass_Li + molar_mass_Fe + molar_mass_P + 4 * molar_mass_O
print(f'Molar mass of LiFePO4: {MM_LiFePO4:.3f} g/mol')

# Integer division (//) and modulo (%)
n_atoms_per_cell = 4 * 7   # 7 atoms in formula unit, 4 formula units per cell
print(f'Atoms per unit cell: {n_atoms_per_cell}')
print(f'Divisible by 4? remainder = {n_atoms_per_cell % 4}')
````

## 1.2 Lists, Tuples, and Dictionaries

Python gives you three everyday containers, and choosing the right one is
mostly about answering "will this collection's contents ever change, and do
I need to look things up by name or by position?"

- **list** — ordered, mutable collection. Use it for a growing series of
  measurements (you can `append` a new reading as it comes in).
- **tuple** — ordered, **immutable** collection (good for fixed data like
  lattice parameters — once a crystal's unit cell is defined, you don't want
  code elsewhere in the notebook to accidentally change one of its values).
- **dict** — key-value store (good for labelled material properties, where
  you want to retrieve a value by its name, e.g. `cathode['capacity_mAh_g']`,
  rather than by remembering "it's the 3rd item in the list").

````python
# ── List: measured capacities from repeat experiments ────────────────────────
capacities_mAh_g = [152.3, 155.1, 149.8, 153.7, 151.2]  # mAh/g

print('All measurements:', capacities_mAh_g)
print('First value:     ', capacities_mAh_g[0])   # indexing from 0
print('Last value:      ', capacities_mAh_g[-1])  # negative index
print('Slice [1:3]:     ', capacities_mAh_g[1:3]) # half-open interval

# Appending and removing
capacities_mAh_g.append(154.0)
print('After append:    ', capacities_mAh_g)
````

````python
# ── Tuple: lattice parameters (a, b, c in Å, alpha, beta, gamma in °) ────────
lattice_LiFePO4 = (10.332, 6.011, 4.695, 90.0, 90.0, 90.0)  # orthorhombic
a, b, c, *angles = lattice_LiFePO4
print(f'a={a} Å, b={b} Å, c={c} Å')
print(f'Angles: {angles}')

# ── Dictionary: material property database entry ─────────────────────────────
cathode = {
    'formula':        'LiFePO4',
    'density_g_cm3':  3.6,
    'capacity_mAh_g': 170,
    'voltage_V':      3.45,
    'structure':      'olivine',
    'is_cathode':     True,
}

print('\nCapacity:', cathode['capacity_mAh_g'], 'mAh/g')
print('Keys:', list(cathode.keys()))
````

## 1.3 Control Flow

**Control flow** is how a program decides what to do next: `if` picks between
alternatives based on a condition, `for` repeats an action once per item in a
collection (e.g. once per sample), and `while` repeats an action until a
condition becomes false (useful when you don't know in advance how many
repetitions you'll need — like simulating a titration until the endpoint is
reached). Almost every data-processing script you write in this course is
built from these three tools, combined with functions (Section 1.4).

````python
# ── if / elif / else ─────────────────────────────────────────────────────────
def classify_capacity(cap_mAh_g):
    """Return a quality label based on specific capacity."""
    if cap_mAh_g >= 200:
        return 'Excellent'
    elif cap_mAh_g >= 150:
        return 'Good'
    elif cap_mAh_g >= 100:
        return 'Acceptable'
    else:
        return 'Poor'

for cap in [230, 165, 120, 85]:
    print(f'{cap:>5} mAh/g  →  {classify_capacity(cap)}')
````

````python
# ── for loop: process a batch of samples ─────────────────────────────────────
samples = [
    {'id': 'S01', 'T_C': 700, 'time_h': 6,  'capacity': 148},
    {'id': 'S02', 'T_C': 750, 'time_h': 6,  'capacity': 161},
    {'id': 'S03', 'T_C': 800, 'time_h': 6,  'capacity': 169},
    {'id': 'S04', 'T_C': 800, 'time_h': 12, 'capacity': 172},
]

print(f'{"Sample":<8} {"T (°C)":<10} {"Time (h)":<12} {"Cap (mAh/g)":<14} {"Quality"}')
print('-' * 55)
for s in samples:
    q = classify_capacity(s['capacity'])
    print(f"{s['id']:<8} {s['T_C']:<10} {s['time_h']:<12} {s['capacity']:<14} {q}")
````

````python
# ── while loop: simulate iterative titration ─────────────────────────────────
volume_added_mL = 0.0
pH = 2.0
dV = 0.5     # increment per step
dpH = 0.18   # pH change per mL NaOH added

print('Simulated NaOH titration')
print(f'{"Volume (mL)":<15} {"pH"}')
while pH < 7.5:
    print(f'{volume_added_mL:<15.1f} {pH:.2f}')
    volume_added_mL += dV
    pH += dpH * dV
print(f'Endpoint reached at {volume_added_mL:.1f} mL (pH = {pH:.2f})')
````

## 1.4 Functions

A function packages up a calculation you'll want to repeat — write the
recipe once, then reuse it by name instead of copy-pasting the same lines
(and risking a typo in one of the copies). Functions make code reusable and
readable.

Use **default arguments** for common values and **keyword arguments** when
calling.

````python
def specific_capacity(charge_mAh, mass_g, efficiency=1.0):
    """
    Calculate specific capacity.

    Parameters
    ----------
    charge_mAh : float   Total charge in mAh
    mass_g     : float   Active material mass in grams
    efficiency : float   Coulombic efficiency (0–1), default 1.0

    Returns
    -------
    float : specific capacity in mAh/g
    """
    return (charge_mAh * efficiency) / mass_g


# Using default efficiency
cap = specific_capacity(charge_mAh=85.6, mass_g=0.5)
print(f'Specific capacity: {cap:.1f} mAh/g')

# Including efficiency
cap_eff = specific_capacity(85.6, 0.5, efficiency=0.97)
print(f'With 97% efficiency: {cap_eff:.1f} mAh/g')
````

````python
# ── Functions returning multiple values ───────────────────────────────────────
import math

def interplanar_spacing_cubic(h, k, l, a):
    """
    d-spacing for cubic crystal: d = a / sqrt(h²+k²+l²)
    Returns d in same units as a, and 2theta in degrees for Cu Kα (λ=1.5406 Å).
    """
    d = a / math.sqrt(h**2 + k**2 + l**2)
    lambda_CuKa = 1.5406  # Å
    sin_theta = lambda_CuKa / (2 * d)
    if abs(sin_theta) <= 1:
        two_theta = math.degrees(2 * math.asin(sin_theta))
    else:
        two_theta = float('nan')
    return d, two_theta

# XRD peaks for gold (FCC, a = 4.078 Å)
a_Au = 4.078
planes = [(1,1,1), (2,0,0), (2,2,0), (3,1,1)]

print(f'{"hkl":<10} {"d (Å)":<12} {"2θ (°)"}')
print('-' * 32)
for hkl in planes:
    d, tt = interplanar_spacing_cubic(*hkl, a=a_Au)
    print(f'{str(hkl):<10} {d:<12.4f} {tt:.3f}')
````

## 1.5 List Comprehensions

List comprehensions are a concise, Pythonic way to build lists from
sequences. `[expression for item in sequence if condition]` reads almost like
English: "for each item in the sequence, if the condition holds, compute the
expression and collect the results into a new list." It replaces a
multi-line `for` loop with `.append()` calls with a single readable line —
useful, but only reach for it when the expression stays simple; a plain
loop is often clearer for anything more complex.

````python
# Temperatures in Celsius → Kelvin
T_C = [25, 100, 200, 400, 600, 800]
T_K = [t + 273.15 for t in T_C]
print('T_K:', T_K)

# Filter: keep only samples with capacity > 160 mAh/g
all_capacities = [148, 161, 169, 172, 145, 158, 163, 141]
good = [c for c in all_capacities if c > 160]
print('Good capacities:', good)

# Combined transformation + filter
# Convert conductance (S) to resistivity (Ω·cm) only for valid (positive) values
conductances_S = [0.12, -0.01, 0.34, 0.0, 0.22, 0.09]  # -0.01 is erroneous
thickness_cm = 0.05
area_cm2 = 1.0
resistivities = [thickness_cm / (g * area_cm2) for g in conductances_S if g > 0]
print('Resistivities (Ω·cm):', [f'{r:.3f}' for r in resistivities])
````

````python
# ── Dictionary comprehension: build property lookup table ────────────────────
cathode_materials = {
    'LCO':     {'capacity': 140, 'voltage': 3.9},
    'LFP':     {'capacity': 170, 'voltage': 3.4},
    'NMC811':  {'capacity': 200, 'voltage': 3.7},
    'LMO':     {'capacity': 120, 'voltage': 4.0},
    'NCA':     {'capacity': 200, 'voltage': 3.7},
}

# Energy density estimate: capacity × voltage
energy_density = {name: d['capacity'] * d['voltage']
                  for name, d in cathode_materials.items()}

print('Estimated gravimetric energy density (mWh/g):')
for name, E in sorted(energy_density.items(), key=lambda x: -x[1]):
    print(f'  {name:<10} {E:.0f} mWh/g')
````

## 1.6 String Formatting (f-strings)

f-strings (Python 3.6+) are the modern standard for formatting output.

````python
sample_id = 'NMC_T800_t12'
capacity  = 198.4267
retention = 0.9341

# Basic f-string
print(f'Sample: {sample_id}')

# Number formatting
print(f'Capacity: {capacity:.2f} mAh/g')     # 2 decimal places
print(f'Capacity: {capacity:8.2f} mAh/g')    # width 8, right-aligned
print(f'Retention: {retention:.1%}')          # percentage
print(f'Retention: {retention:.4e}')          # scientific notation

# Building a report line
def report_line(sid, cap, ret):
    flag = '✓' if ret > 0.9 else '✗'
    return f'{flag} {sid:<20} {cap:>8.2f} mAh/g   {ret:.1%} retention'

print()
print(report_line('NMC_T800_t12', 198.4, 0.934))
print(report_line('NMC_T700_t06', 155.1, 0.871))
````
