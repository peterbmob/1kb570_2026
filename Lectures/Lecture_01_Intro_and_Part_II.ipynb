| *"Write the code to run a two-sample t-test on this data"* | *"I think I need a two-sample t-test here — what should I check before trusting one?"* |

The left column gets you a working cell today and a blank page in the

## Why Generated (Simulated) Data?

Every case study in this course — ZnO grain size, NMC cathode capacity, LiFePO4 batches, TiO2 photocatalysis — starts from a **known true relationship plus random noise**. That is deliberate:

1. **You can check whether a method actually recovers the truth** — a t-test, a PCA, a factorial design — because you built the true answer in yourself. Real data never lets you do that.
2. **It's fully reproducible** — same seed, same numbers, same discussion, for every student, every time
3. **It's safe and fast** — no lab time, no reagent cost; try 10,000 experiments before running one real one
4. **It mirrors real practice** — this is exactly what a statistician calls a *Monte Carlo check*


The core meta-lesson for this segment. Consider asking the room: 'has anyone wondered
why every example uses np.random instead of a CSV of real measurements?' before
revealing these four reasons. This pattern continues through Parts III, IV and V.

## A Synthetic Dataset, Live

Simulating 100 charge-discharge cycles of a battery — exponential capacity fade plus measurement noise. **Run the cell below live.**


````python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(4)          # <- try changing this seed and re-running!
cycle_numbers = np.arange(1, 101)
capacity_0 = 170.0     # mAh/g
fade_rate = 0.0015     # capacity loss per cycle

capacity = capacity_0 * np.exp(-fade_rate * cycle_numbers) + rng.normal(0, 0.5, 100)

plt.figure(figsize=(8, 4.5))
plt.plot(cycle_numbers, capacity, '.', color='steelblue')
plt.xlabel('Cycle number'); plt.ylabel('Capacity (mAh/g)')
plt.title('Simulated battery capacity fade')
plt.show()

print(f'Cycle 1:   {capacity[0]:.2f} mAh/g')
print(f'Cycle 100: {capacity[99]:.2f} mAh/g')
````

**Try it live:** change `rng = np.random.default_rng(7)` to a different number and re-run — same code, different noise, *same underlying truth*. This one cell already combines three Part I tools: **NumPy** (generates/computes the array), **pandas** (would label it), **Matplotlib** (renders it).


# Collaboration: Git & GitHub

## The Problem: `analysis_v2_final_REALLY.ipynb`

Renaming files by hand fails, and gets worse the longer a project runs:
- No record of **why** something changed
- No way to see exactly what changed between two versions without eyeballing both
- No safe way for two people to edit the same file at once without overwriting each other

**Version control (git) fixes all three at once** — every commit records exactly what changed, who, and why; the complete history is always available; two people's independent changes can usually be combined automatically.


## A Commit Is Just a Labelled Snapshot

A tiny in-memory stand-in for git's commit history — the same mental model as the real thing.


````python
import hashlib
from datetime import datetime, timedelta

class ToyRepo:
    '''A tiny, in-memory stand-in for git's commit history.'''
    def __init__(self):
        self.commits = []          # chain of snapshots, oldest first
        self.working_files = {}    # current file contents

    def _commit_id(self, message, timestamp):
        raw = f'{message}{timestamp}{len(self.commits)}'.encode()
        return hashlib.sha1(raw).hexdigest()[:7]

    def commit(self, message, author='you'):
        parent = self.commits[-1]['id'] if self.commits else None
        timestamp = datetime.now() + timedelta(seconds=len(self.commits))
        snapshot = {'id': self._commit_id(message, timestamp), 'parent': parent,
                    'message': message, 'author': author, 'timestamp': timestamp,
                    'files': dict(self.working_files)}
        self.commits.append(snapshot)
        return snapshot['id']

    def log(self):
        for c in reversed(self.commits):
            print(f"{c['id']}  {c['timestamp']:%Y-%m-%d %H:%M:%S}  {c['author']:<8} {c['message']}")


repo = ToyRepo()
repo.working_files['analysis.py'] = "print('load data')"
repo.commit('Initial analysis script')

repo.working_files['analysis.py'] += "\nprint('clean data')"
repo.commit('Add data cleaning step')

repo.log()
````

Nothing is ever overwritten — an earlier snapshot is always reachable through the chain. Real commands: `git init` · `git add` · `git commit -m "message"` · `git log`.


## Branches: Try an Idea Without Breaking What Works

Keep a stable line of commits untouched (`main`) while trying something risky on a separate line.


````python
class BranchingRepo(ToyRepo):
    def __init__(self):
        super().__init__()
        self.branches = {'main': None}   # branch name -> commit id it points to
        self.current_branch = 'main'

    def commit(self, message, author='you'):
        commit_id = super().commit(message, author)
        self.branches[self.current_branch] = commit_id
        return commit_id

    def branch(self, name):
        self.branches[name] = self.branches[self.current_branch]

    def checkout(self, name):
        self.current_branch = name

    def status(self):
        print(f"On branch '{self.current_branch}'")
        for name, cid in self.branches.items():
            marker = ' <-- you are here' if name == self.current_branch else ''
            print(f"  {name}: {cid}{marker}")


brepo = BranchingRepo()
brepo.working_files['model.py'] = "model = 'MLR'"
brepo.commit('Baseline: linear regression model')

brepo.branch('try-random-forest')     # fork off from the current commit
brepo.checkout('try-random-forest')
brepo.working_files['model.py'] = "model = 'RandomForest'"
brepo.commit('Experiment: swap in a Random Forest')

brepo.status()
````

`main` still points at the safe baseline commit — the experiment cannot break it, no matter how badly it goes. Same discipline as testing a risky package upgrade in a throwaway conda environment first (Notebook 6).

## Data Without Metadata Is (Almost) Useless

`182, 195, 188, 201, 179, 193`

What is this? Hardness values? In what units — Vickers, Brinell, Rockwell? Which sample? Which instrument? Measured when, by whom?

### A bare number is scientifically worthless without its context

## The FAIR Principles

| | Principle | The plain-language question |
|---|---|---|
| **F** | Findable | Can a person or machine locate this dataset at all? |
| **A** | Accessible | Can it actually be fetched — even under conditions? |
| **I** | Interoperable | Can it combine with other data/tools with no manual translation? |
| **R** | Reusable | Is there enough context to reuse it correctly? |

**FAIR is a checklist, not a synonym for "open."** Data can be FAIR and still access-controlled or paywalled.


## Metadata in Python: Dictionaries and JSON

A dictionary is the natural structure for metadata — it's inherently nested. JSON is the standard interchange format (human-readable, near-lossless, what NOMAD uses under the hood).


````python
import json

cathode_record = {
    'identifier': 'urn:uuid:1234',
    'title': 'LiFePO4 cathode, solid-state synthesis batch 07',
    'licence': 'CC-BY-4.0',
    'creator': 'A. Lindqvist',
    'created': '2025-01-15',
    'method': 'solid-state reaction, 2-step calcination',
    'data': {'temperature_C': 750, 'time_h': 6, 'capacity_mAh_g': 161.4},
}

def fair_check(record):
    '''Which required FAIR-relevant fields, if any, are missing?'''
    required = ['identifier', 'title', 'licence', 'creator', 'created', 'method']
    missing = [k for k in required if k not in record]
    return not missing, missing

ok, missing = fair_check(cathode_record)
print('FAIR check passed:', ok)

# JSON round-trip: save, then load back
record_json = json.dumps(cathode_record, indent=2)
reloaded = json.loads(record_json)
print('Round-trip identical:', reloaded == cathode_record)
print(record_json)
````

*This exact record reappears across the rest of today — rebuilt as an ontology individual, SQL rows, MongoDB documents, and a NOMAD-style archive.*


## The Problem: Words Are Not Reliable Identifiers

**Synonymy** — same concept, different words
- "yield strength" / "0.2% proof stress" / σy — a naive string filter misses the aliases

**Homonymy** — same word, different concepts
- "annealing" in metallurgy vs. molecular biology
- "conductivity" without a qualifier — electrical, or thermal?

A column header is just a string — it cannot disambiguate on its own.


## Controlled Vocabulary → Taxonomy → Ontology

| Level | What it provides | Example | Limitation |
|---|---|---|---|
| Controlled vocabulary | A flat, fixed term list | `{as-cast, annealed, quenched}` | No structure |
| Taxonomy | Vocabulary + an is-a hierarchy | Linnaean classification | One relation type only |
| **Ontology** | Taxonomy + relations, properties, formal logic | **EMMO** | Real upfront modelling effort |


## EMMO — The European Materials & Modelling Ontology

A **foundational, top-level ontology** for physical science and engineering.

- Developed/maintained by the European Materials Modelling Council (EMMC)
- Open, versioned, at a persistent ID (`w3id.org/emmo`), CC-BY-4.0
- Every concept ultimately derives from one root class: `Item`
- Domain ontologies build on top of it — e.g. **BattINFO** for FAIR battery data
- NOMAD increasingly annotates its own schemas with EMMO concepts


## The Key Insight: A Measurement Is Not the Property

**A material — Crystal**
`Crystal → CrystallineMaterial → Material → OrdinaryMatter → PhysicalObject → … → Item`

**A quantity — SpecificHeatCapacity**
`SpecificHeatCapacity → Sign → SemioticEntity → Semiotics → … → Item`

*(Real EMMO ancestor chains, walked live via SPARQL against the actual ontology.)*

"188 HV" is a **Sign** representing a real hardness, not the hardness itself.


## Try It: A Toy Ontology, Live

Building the same idea in miniature with `owlready2` — the same building blocks EMMO itself is built from.


````python
from owlready2 import get_ontology, Thing, ObjectProperty

onto = get_ontology("http://example.org/toy_emmo.owl")

with onto:
    class Item(Thing): pass
    class PhysicalEntity(Item): pass
    class Quantity(Item): pass
    class MaterialObject(PhysicalEntity): pass
    class CathodeMaterial(MaterialObject): pass
    class ThermodynamicTemperature(Quantity): pass

    class hasQuantitativeProperty(ObjectProperty):
        domain = [MaterialObject]
        range = [Quantity]

batch_07 = CathodeMaterial("LiFePO4_batch_07")
T_synthesis = ThermodynamicTemperature("T_synthesis_batch_07")
batch_07.hasQuantitativeProperty.append(T_synthesis)

print("Is batch_07 a MaterialObject?", isinstance(batch_07, MaterialObject))
print("Ancestors of CathodeMaterial:", [c.name for c in CathodeMaterial.ancestors()])
````

## Beyond the Hierarchy: What Makes an Ontology *Do* Things

## Two Data Models: Rows vs. Documents

| | Relational (SQL) | Document (NoSQL) |
|---|---|---|
| Shape | Fixed-column tables of rows | Nested, JSON-like documents |
| Schema | Enforced upfront | Flexible, per-document |
| Relationships | Foreign keys + JOIN | Usually embedded |
| Query language | SQL | Method-call API |
| Systems | PostgreSQL, MySQL, SQLite | MongoDB, CouchDB |
| Good fit for | Strictly tabular, strict integrity | Heterogeneous, nested data |


## SQL, Live

A small relational schema — `samples` and `results`, linked by a foreign key. Watch what happens when a `NOT NULL` constraint is violated.


````python
import sqlite3
import pandas as pd

conn = sqlite3.connect(':memory:')
conn.execute('''CREATE TABLE samples (
    sample_id TEXT PRIMARY KEY, material TEXT, method TEXT NOT NULL
)''')
conn.execute('''CREATE TABLE results (
    sample_id TEXT PRIMARY KEY, capacity_mAh_g REAL,
    FOREIGN KEY(sample_id) REFERENCES samples(sample_id)
)''')

samples = [('batch_05', 'LiFePO4', 'solid-state'), ('batch_06', 'LiFePO4', 'sol-gel'),
           ('batch_07', 'LiFePO4', 'solid-state')]
conn.executemany('INSERT INTO samples VALUES (?,?,?)', samples)
results = [('batch_05', 158.2), ('batch_06', 163.9), ('batch_07', 161.4)]
conn.executemany('INSERT INTO results VALUES (?,?)', results)
conn.commit()

df = pd.read_sql_query('''
    SELECT s.sample_id, s.method, r.capacity_mAh_g
    FROM samples s JOIN results r ON s.sample_id = r.sample_id
''', conn)
print(df)

# Schema enforcement, live: try to insert a sample with no method
try:
    conn.execute("INSERT INTO samples VALUES ('batch_08', 'LiFePO4', NULL)")
except sqlite3.IntegrityError as e:
    print('\nRejected:', e)
````

## What Is NOMAD?

An **open-source, FAIR data infrastructure** for materials science (`nomad-lab.eu`).

- Developed by **FAIRmat** — part of Germany's National Research Data Infrastructure (NFDI)
- Started for computational/DFT/simulation data
- Has grown **Electronic Lab Notebook (ELN)** support for experimental data

**NOMAD Oasis**: the same software, run as your own private instance — for unpublished results or embargoed data, speaking the same data model as public NOMAD.


## Anatomy of a NOMAD Entry

| Layer | What it is | Analogy from earlier today |
|---|---|---|
| Raw data | The original files, untouched | The instrument's raw export |
| Archive | Parsed into NOMAD's structured schema | The FAIR `cathode_record` dict — now machine-validated |
| Metadata | Searchable summary fields | The F / A / R fields from `fair_check` |


## Where EMMO Fits In, Live

NOMAD schemas attach an ontology **concept IRI** to individual quantities, so differently-named fields in different labs can still be recognised as the same concept.


````python
nomad_style_entry = {
    'metadata': {'upload_id': 'upload_2025_01_cathodes', 'entry_id': 'entry_LiFePO4_batch07',
                 'authors': ['A. Lindqvist'], 'licence': 'CC-BY-4.0'},
    'archive': {'data': {
        'm_def': 'CathodeSynthesisEntry',
        'synthesis': {'method': 'solid-state reaction'},
        'results': {'capacity_mAh_g': 161.4,
                    'quantity_kind_iri': 'emmo:ElectricCharge'},   # <- the EMMO tag
    }},
}

def validate_entry(entry, schema):
    '''A minimal structural validator -- same spirit as fair_check, checking
    structure instead of FAIR-ness.'''
    for key, expected_type in schema.items():
        node = entry
        for part in key.split('.'):
            if part not in node:
                return False, f'missing {key}'
            node = node[part]
        if not isinstance(node, expected_type):
            return False, f'{key} has wrong type'
    return True, None

schema = {
    'archive.data.synthesis.method': str,
    'archive.data.results.capacity_mAh_g': (int, float),
}
print(validate_entry(nomad_style_entry, schema))
````
