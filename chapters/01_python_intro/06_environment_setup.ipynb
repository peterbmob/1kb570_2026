## 6.1 Why a Dedicated Environment?

Every computer that has ever run Python already has *a* Python installed
somewhere — so why not just use that? Because two different projects almost
never want the exact same package versions, and installing everything into
one shared, system-wide Python turns that disagreement into a fight neither
project can win. This course's own `requirements.txt` has a real example of
exactly this problem, one line long:

```
scikit-learn>=1.5,<1.6   # factor_analyzer 0.5.1 is incompatible with sklearn>=1.6
```

Part IV's Factor Analysis notebook depends on a package (`factor_analyzer`)
that was never updated for scikit-learn's newest release — install the
latest scikit-learn system-wide for some *other* project, and this course's
own Factor Analysis notebook silently breaks. An **environment** (created by
`conda` or Python's built-in `venv`) sidesteps this entirely: it is a named,
self-contained set of installed packages at pinned versions, completely
separate from your system Python and from every other project's environment.
Different projects, different environments, no fights — this is precisely
the same "pin your dependencies" discipline behind every `requirements.txt`
and `environment.yml` you will ever see in a research codebase, and it is
worth adopting from your very first project, not just this course.

## 6.2 Installing Anaconda or Miniconda Locally

**Anaconda** and **Miniconda** are two installers for the same underlying
tool, `conda`: a package and environment manager. Anaconda bundles conda with
roughly 250 pre-installed scientific packages and a graphical launcher
(Anaconda Navigator) — a multi-gigabyte download convenient for someone who
wants everything available immediately. Miniconda installs just Python and
conda itself (a few hundred MB) and lets you install only what a given
project needs — the better default for a course like this one, since
`requirements.txt` already specifies exactly what is needed, and you avoid
downloading 250 packages you will never import.

**Installing Miniconda** (Windows, macOS, or Linux):
1. Download the installer for your OS from
   [docs.conda.io/en/latest/miniconda.html](https://docs.conda.io/en/latest/miniconda.html).
2. Run it, accepting the defaults unless you have a specific reason not to.
3. Open a **new** terminal (PowerShell on Windows, Terminal on macOS/Linux) —
   `conda` is only on your `PATH` in terminals opened *after* installation —
   and confirm it worked:

```bash
conda --version
```

## 6.3 Reproducing This Course's Environment

Every notebook in this book was executed inside one specific environment,
defined by the repository's own `requirements.txt` (the same file
[intro.md](../../intro.md) already told you about). Creating a matching
environment is three commands: make a new, empty conda environment, activate
it, then install every package `requirements.txt` lists into it.

```bash
conda create -n doe-course python=3.11
conda activate doe-course
pip install -r requirements.txt
```

Notice the third command is `pip`, not `conda install` — conda manages the
Python *interpreter* and the isolated environment itself perfectly well,
while `pip` (Python's own package installer, also used with the `!` prefix
inside notebooks, e.g. Notebook 5's live NOMAD queries) installs the
packages the environment actually needs. Mixing the two like this — conda
for the environment, pip for the packages — is extremely common in practice
and exactly what this line does.

Once installed, launching **JupyterLab** gives you the same browser-based
notebook interface you have been reading this whole book in:

```bash
conda activate doe-course
jupyter lab
```

## 6.4 Zero-Install: Google Colab

**Google Colab** (`colab.research.google.com`) runs Jupyter notebooks
entirely on Google's servers, in your browser, with no install at all — you
only need a Google account. It is the fastest way to get *any* notebook
running with zero local setup, and it includes free (rate-limited) access to
a GPU, useful for Part VI's PyTorch notebook (Notebook 20) if your own
laptop has none.

**Opening a notebook**: `File → Upload notebook` for a `.ipynb` file on your
computer, or `File → Open notebook → GitHub` and paste a repository URL —
exactly the workflow Notebook 7 (next) will make you comfortable with.

**Installing this course's packages**: Colab's own environment does not
match `requirements.txt` out of the box. The `!` prefix runs a plain shell
command from inside a code cell (you first saw this idea, without the cloud
context, back in Notebook 5's live NOMAD queries) — so the very first cell
of a Colab-run notebook for this course would be:

```text
!pip install -r requirements.txt
```

**The catch — persistence**: a Colab session is a temporary virtual machine.
Close the tab (or leave it idle for a while) and everything not explicitly
saved elsewhere — installed packages, any files you created — disappears
with it. For anything you want to keep, mount your Google Drive first:

```python
from google.colab import drive
drive.mount('/content/drive')
```

Good for: trying a notebook with no local install at all, occasional GPU
access, sending a colleague a working link instead of a file attachment.
Not good for: long-running jobs, anything that must persist unattended, or
work you cannot afford to lose to a disconnected runtime.

## 6.5 Uppsala's Own Cluster: UPPMAX

**UPPMAX** (Uppsala Multidisciplinary Center for Advanced Computational
Science) is Uppsala University's own research computing centre — as a
1KB570 student, you can normally request an account through
[SUPR](https://supr.naiss.se) (the Swedish national allocation system) via
your course or research group. It exists for exactly the situation Colab and
a laptop both struggle with: a job that needs more memory or CPU time than
either can comfortably give you — a large factorial design across many
factors (Part V), or training several models over a wide hyperparameter grid
(Part VI) are realistic examples from this course specifically.

**Getting to a notebook on UPPMAX**, in increasing order of how much of the
cluster you are using:
1. **UPPMAX's JupyterHub/JupyterLab web portal** (when available on your
   cluster) — the closest experience to this book: open a browser, get a
   notebook, no terminal required for basic use.
2. **An interactive terminal session** via SSH or UPPMAX's ThinLinc remote
   desktop, using the `module` system to load pre-installed software
   (`module load python/3.11` or similar — exact module names are
   cluster-specific and change over time).
3. **A batch job submitted through SLURM**, the cluster's job scheduler, for
   long, unattended runs that do not need you watching a notebook live.

**Your own environment on UPPMAX**: you will not have administrator rights
on a shared cluster, but you do not need them — the same `conda
create`/`pip install -r requirements.txt` pattern from Section 6.3 works
identically once you are logged in, with one practical wrinkle: shared
clusters enforce **storage quotas** on your home directory, and a full
scientific Python environment can be several GB, so UPPMAX documentation
generally recommends building environments in your allocated **project
storage** rather than your home directory.

UPPMAX's own procedures (account requests, current module names, JupyterHub
availability) change over time and are documented properly at
[uppmax.uu.se](https://www.uppmax.uu.se) — treat this section as
orientation for *why* and *when* you would reach for UPPMAX, and go to their
documentation (or your course coordinator) for the exact, current steps.

## 6.6 Choosing the Right Option

None of these three is simply "the best" — each trades off setup effort
against compute power and persistence differently.

| | Local (Anaconda/Miniconda) | Google Colab | UPPMAX |
|---|---|---|---|
| Setup effort | One install, once | None — just a browser | Account request + first-time setup |
| Cost | Free (your own hardware) | Free tier is generous; limits apply | Free for eligible university users |
| Compute power | Whatever your machine has | Free CPU; free but rate-limited GPU | Large-scale CPU/GPU, shared with other users |
| Persistence | Permanent, entirely yours | Ephemeral unless saved to Drive | Permanent, quota-limited storage |
| Offline use | Yes | No — needs a connection | No — needs a connection |
| Best for | Day-to-day coursework, full control | Quick trials, sharing a runnable link, occasional GPU | Large jobs beyond a laptop's capacity |

A very common real pattern: develop and debug on a small subset of your data
locally or in Colab, where iterating is fast, then move the finished,
working analysis to UPPMAX only for the parts that actually need serious
compute — exactly the "get it right small, then scale up" workflow DoE
itself (Part V) is built around.

````python
import sys
import platform

print(f'Python version: {sys.version.split()[0]}')
print(f'Platform:       {platform.platform()}')
print(f'Executable:     {sys.executable}')
````

If the Python version above is 3.10 or newer, you are set for every
notebook in this course. `sys.executable` shows *which* Python is actually
running this cell — worth checking any time an `import` mysteriously fails
even though you "definitely installed that package": it is often installed
into a *different* environment than the one your notebook is using.
